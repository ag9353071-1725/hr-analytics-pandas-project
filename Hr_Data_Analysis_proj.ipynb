```python
import pandas as pd

data = pd.read_csv(r"D:\Hr dataset\HR_Dataset.csv")
```

```python
# check anu null value is in HR dataset
data.isnull().sum()  #No missing value found in dataset
```

```python
#Avg salary per department
avg_sal= data.groupby('Department')['Monthly_Income'].mean()
print(avg_sal)
```

```python
#Avg salary per department in asc
avg_salary=data.groupby('Department')['Monthly_Income'].mean().round(2).sort_values(ascending = False) #HR department is the highest avg paid salary
print(avg_salary)
```

```python
Highest_salary = data.loc[data['Monthly_Income'].idxmax(),['Employee_ID','Monthly_Income']] #IT GIVES HIGHEST SALARY BY EMPLOYEEE_ID
print(Highest_salary)
```

```python
#top 10 highest paid salary  
top_10 = data[['Employee_ID','Monthly_Income']].sort_values(by='Monthly_Income',ascending = False).head(10)
print(top_10.to_string(index=False))
```

```python
#salary greater than 10000
emp = data.loc[data['Monthly_Income'] >10000,['Employee_ID','Monthly_Income']]
print(emp.to_string(index=False))
print('Number of employee earning more than 10000 :' ,len(emp))

#output
# There are 272 employee who earning morethan 10000
```

```python
data['Attrition'].value_counts()
#output
#There are 3400 employee in the company but 1724 employee left the company and 1676 employee still in the company
```

```python
attrition_rate = (data['Attrition'].value_counts(normalize=True)*100).round(2)
print(attrition_rate)
#output
#it gives attrition rate that 50.71% employee left the company and 49.29% employee still work in company 
#It can affect company growth future
```

```python
pd.crosstab(data['Department'],data['Attrition'])

#"A cross-tabulation analysis was performed between Department and Attrition. The Sales department recorded the highest employee attrition (359 employees), while the IT department recorded the lowest attrition (329 employees). Attrition levels were relatively similar across all departments, indicating that employee turnover is distributed fairly evenly throughout the organization."
```
